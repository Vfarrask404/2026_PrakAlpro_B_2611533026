## 2.1 Dasar Teori
### 2.1.1 Pengertian Percabangan (Conditional Statement)
Percabangan atau conditional statement adalah struktur kontrol dalam pemrograman yang memungkinkan program untuk mengambil keputusan dan mengeksekusi blok kode tertentu berdasarkan suatu kondisi (boolean expression) yang bernilai True atau False. Dalam bahasa pemrograman Python, struktur percabangan diimplementasikan menggunakan kata kunci if, elif, dan else.

Berbeda dengan bahasa pemrograman lain seperti C++ atau Java yang menggunakan tanda kurung kurawal {} untuk menandai blok kode, Python menggunakan indentasi (biasanya 4 spasi) untuk menentukan cakupan (scope) dari suatu blok percabangan. Kesalahan indentasi pada Python akan menyebabkan error, karena indentasi bersifat wajib, bukan sekadar gaya penulisan.

### 2.1.2 Struktur if Tunggal
Struktur `if` tunggal digunakan ketika program hanya perlu memeriksa satu kondisi. Jika kondisi tersebut bernilai True, maka blok kode di dalamnya akan dieksekusi. Jika kondisi bernilai False, blok kode tersebut akan dilewati begitu saja.

### 2.1.3 Struktur if-else
Struktur `if-else` digunakan ketika terdapat dua kemungkinan hasil dari suatu kondisi: jika kondisi bernilai True, maka blok `if` yang dijalankan; jika kondisi bernilai False, maka blok `else` yang dijalankan. Struktur ini memastikan salah satu dari dua blok kode pasti akan dieksekusi.

### 2.1.4 Multi if (If Bertingkat)
`Multi if`, atau sering disebut `nested if`, adalah penggunaan pernyataan if di dalam blok if atau else lainnya. Struktur ini digunakan ketika suatu kondisi memiliki sub-kondisi yang harus diperiksa lebih lanjut setelah kondisi utama terpenuhi. `Nested if` memungkinkan pengecekan kondisi secara bertingkat/berlapis.

### 2.1.5 Struktur if-elif-else
Struktur `if-elif-else` digunakan ketika terdapat lebih dari dua kemungkinan kondisi yang harus diperiksa secara berurutan. Kata kunci elif merupakan singkatan dari "else if", yang memungkinkan program memeriksa kondisi tambahan apabila kondisi sebelumnya tidak terpenuhi. Python akan mengevaluasi kondisi dari atas ke bawah, dan begitu satu kondisi terpenuhi (True), blok kode yang sesuai akan dijalankan dan sisa kondisi lainnya tidak akan diperiksa lagi.

### 2.2.1 Program "if1_2611533026.py"
Program ini dibuat untuk membantu pengguna mengetahui status kelulusan berdasarkan nilai Indeks Prestasi Kumulatif (IPK) yang dimilikinya. Pengguna diminta memasukkan nilai IPK, kemudian program akan memeriksa apakah nilai tersebut memenuhi syarat kelulusan dengan predikat sangat memuaskan, yaitu di atas 2.75. Program ini menerapkan konsep struktur percabangan (conditional statement) dalam bahasa pemrograman Python, khususnya penggunaan pernyataan if tunggal, untuk menentukan alur eksekusi program berdasarkan suatu kondisi tertentu.

</div>

```python
ipk_3026 = float(input("Input ipk anda = "))

if ipk_3026 > 2.75 :
    print("anda lulus sangat memuaskan dengan ipk " + str(ipk_3026))

print("program selesai")
```

<div style="font-family: 'Times New Roman', Times, serif; font-size: 18px;">

**Penjelasan :**

1. Baris `ipk_3026 = float(input("Input ipk anda = "))` meminta pengguna memasukkan nilai IPK, lalu mengubahnya dari teks menjadi angka desimal menggunakan `float()`, dan menyimpannya ke dalam variabel `ipk_3026`.
2. Baris `if ipk_3026 > 2.75 :` memeriksa apakah nilai IPK yang dimasukkan lebih besar dari 2.75.
3. Jika kondisi tersebut bernilai `True`, baris `print("anda lulus sangat memuaskan dengan ipk " + str(ipk_3026))` akan dijalankan, mencetak pesan kelulusan beserta nilai IPK-nya (nilai diubah dulu menjadi teks dengan `str()` agar bisa digabung dengan string lain).
4. Jika kondisi tersebut bernilai `False`, baris di dalam blok `if` dilewati begitu saja, karena tidak ada `else` yang menangani kondisi tersebut.
5. Baris terakhir, `print("program selesai")`, tidak berada di dalam blok `if` (tanpa indentasi tambahan), sehingga baris ini akan selalu dijalankan apa pun hasil pengecekan kondisi sebelumnya.

### 2.2.2 Program "if2_2611533026.py"

Jika tidak memenuhi syarat tersebut, program akan menampilkan pesan bahwa pengguna tidak lulus. 4. Jika kondisi tersebut bernilai `False`, program akan menjalankan blok `else`, yaitu baris `print("Anda tidak lulus")`, yang mencetak pesan bahwa pengguna tidak lulus.
5. Baris terakhir, `print("program selesai")`, tidak berada di dalam blok `if` maupun `else` (tanpa indentasi tambahan), sehingga baris ini akan selalu dijalankan apa pun hasil pengecekan kondisi sebelumnya.

### 2.2.3 Program "multi_if1_2611533026.py"
Program ini dibuat untuk menentukan apakah seseorang boleh mengendarai motor berdasarkan dua faktor, yaitu umur dan kepemilikan SIM C. Pengguna diminta memasukkan umur serta status kepemilikan SIM C (y/t), kemudian program akan memeriksa kombinasi dari kedua data tersebut untuk menghasilkan salah satu dari empat kemungkinan pesan. Program ini menerapkan konsep struktur percabangan (conditional statement) dalam bahasa pemrograman Python, khususnya penggunaan beberapa pernyataan if sejajar yang dikombinasikan dengan operator logika and, untuk menentukan alur eksekusi program berdasarkan lebih dari satu kondisi sekaligus.
</div>

```python
umur_3026 = int(input("input umur anda : "))
sim_3026 = input("Apakah anda udah punya SIM C (y/t): ") [0]

if umur_3026 >= 17 and sim_3026 == 'y' :
    print("anda sudah dewasa dan boleh bawa motor")

if umur_3026 >= 17 and sim_3026 != 'y' :
    print("anda sudah dewasa tetapi tidak boleh bawa motor")

if umur_3026 < 17 and sim_3026 == 'y' :
    print("anda belum cukup umur punya SIM")

if umur_3026 < 17 and sim_3026 != 'y' :
    print ("anda belum cukup umur bawa motor")
```

<div style="font-family: 'Times New Roman', Times, serif; font-size: 18px;">

**Penjelasan :**
1. Baris `umur_3026 = int(input("input umur anda : "))` meminta pengguna memasukkan umur, lalu mengubahnya dari teks menjadi bilangan bulat menggunakan `int()`, dan menyimpannya ke dalam variabel `umur_3026`.
2. Baris `sim_3026 = input("Apakah anda udah punya SIM C (y/t): ")[0]` meminta pengguna memasukkan jawaban (y/t), lalu mengambil karakter pertama dari jawaban tersebut menggunakan indexing `[0]`, dan menyimpannya ke dalam variabel `sim_3026`.
3. Baris `if umur_3026 >= 17 and sim_3026 == 'y' :` memeriksa apakah umur sudah 17 tahun ke atas **dan** sudah punya SIM C, menggunakan operator logika `and` yang mengharuskan kedua kondisi terpenuhi bersamaan. Jika `True`, program mencetak pesan bahwa pengguna sudah dewasa dan boleh bawa motor.
4. Baris `if umur_3026 >= 17 and sim_3026 != 'y' :` memeriksa apakah umur sudah 17 tahun ke atas **namun** belum punya SIM C (menggunakan operator `!=` untuk "tidak sama dengan"). Jika `True`, program mencetak pesan bahwa pengguna sudah dewasa tetapi tidak boleh bawa motor.
5. Baris `if umur_3026 < 17 and sim_3026 == 'y' :` memeriksa apakah umur masih di bawah 17 tahun **namun** sudah punya SIM C. Jika `True`, program mencetak pesan bahwa pengguna belum cukup umur untuk punya SIM.
6. Baris `if umur_3026 < 17 and sim_3026 != 'y' :` memeriksa apakah umur masih di bawah 17 tahun **dan** belum punya SIM C. Jika `True`, program mencetak pesan bahwa pengguna belum cukup umur untuk bawa motor.
7. Keempat pernyataan `if` ini bersifat **independen** (bukan `if-elif-else`), sehingga setiap `if` akan tetap dievaluasi satu per satu dari atas ke bawah tanpa memedulikan hasil `if` sebelumnya — namun karena keempat kondisinya saling eksklusif (tidak mungkin dua kondisi terpenuhi sekaligus), pada akhirnya hanya satu pesan saja yang akan tercetak.
8. Keempat kata kunci `if` tersebut wajib ditulis **sejajar** (berada pada posisi indentasi yang sama, yaitu tanpa indentasi/rata kiri). Ini penting karena Python menggunakan indentasi untuk menentukan struktur blok kode — jika salah satu `if` ditulis dengan indentasi yang berbeda dari yang lain, Python akan menganggapnya sebagai bagian dari blok kode yang lain (misalnya dianggap bersarang/nested di dalam `if` sebelumnya) atau bahkan menyebabkan error `IndentationError`.

### 2.2.4 Program "multi_if2_2611533026.py"
Program ini dibuat untuk menghitung besarnya diskon dan total pembayaran belanja berdasarkan tiga syarat, yaitu nominal belanja di atas Rp1.000.000, status keanggotaan (member), dan keabsahan kode promo. Setiap syarat yang terpenuhi akan menambahkan persentase diskonnya masing-masing, sehingga total diskon yang didapatkan bisa merupakan gabungan dari beberapa syarat sekaligus. Program ini menerapkan konsep struktur percabangan (*conditional statement*) dalam bahasa pemrograman Python, khususnya penggunaan beberapa pernyataan `if` sejajar yang bersifat independen, untuk menghitung akumulasi diskon berdasarkan beberapa kondisi yang dapat terpenuhi secara bersamaan.
</div>

```python
#program menghitung diskon belanja

#input user
total_belanja_3026 = float(input("Masukkan total belanja (Rp) : "))

#input status member (mengecek apakah user mengetik 'y" atau 'ya')
input_member_3026 = input("Apakah anda member? (y/t): ").strip().lower()
is_member_3026 = input_member_3026 in ["y", "ya"]

# input status kode promo (mengecek apakah user mengetik 'y' atau 'ya)
input_promo_3026 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3026 = input_promo_3026 in ["y", "ya"]

total_diskon_persen_3026 = 0

if total_belanja_3026 > 1000000 :
    total_diskon_persen_3026 += 10 # diskon belanja besar

if is_member_3026 :
    total_diskon_persen_3026 += 5 #diskon member

if kode_promo_valid_3026 :
    total_diskon_persen_3026 += 15 #diskon voucher

#menghitung nominal diskon dan total bayar
nominal_diskon_3026 = total_belanja_3026 * (total_diskon_persen_3026 / 100)
total_bayar_3026 = total_belanja_3026 - nominal_diskon_3026

#output hasil
print("\n---Rincian Pembayaran---")
print(f"total diskon :  {total_diskon_persen_3026}% (Rp {nominal_diskon_3026 :,.0f})")
print(f"Total Bayar : Rp {total_bayar_3026:,.0f}")

print(f"Total diskon yang anda dapatkan :  {total_diskon_persen_3026}%")
#output :total diskon yang anda dapatkan : 30% jika belanja > 1 juta, member, dan kode promo valid
```

<div style="font-family: 'Times New Roman', Times, serif; font-size: 18px;">

**Penjelasan :**
1. Baris `total_belanja_3026 = float(input("Masukkan total belanja (Rp) : "))` meminta pengguna memasukkan total belanja, lalu mengubahnya menjadi angka desimal dengan `float()`.
2. Baris `input_member_3026 = input(...).strip().lower()` meminta pengguna mengetik status member, lalu `.strip()` menghapus spasi di awal/akhir dan `.lower()` mengubah semua huruf menjadi huruf kecil, agar penulisan seperti "Y", " ya ", atau "YA" tetap terbaca sama.
3. Baris `is_member_3026 = input_member_3026 in ["y", "ya"]` memeriksa apakah jawaban tadi termasuk dalam daftar `["y", "ya"]`. Hasilnya berupa nilai boolean (`True`/`False`) yang langsung disimpan ke variabel `is_member_3026`.
4. Baris `input_promo_3026` dan `kode_promo_valid_3026` bekerja dengan cara yang sama seperti poin 2 dan 3, tetapi untuk memeriksa apakah kode promo yang dimasukkan valid.
5. Baris `total_diskon_persen_3026 = 0` membuat variabel penampung persentase diskon dan mengawalinya dengan nilai 0.
6. Tiga blok `if` berikutnya masing-masing memeriksa satu syarat diskon secara terpisah (belanja di atas Rp1.000.000, status member, dan kode promo valid). Karena ketiganya berupa `if` sejajar (bukan `elif`), setiap syarat yang terpenuhi akan menambahkan persentasenya sendiri ke `total_diskon_persen_3026` menggunakan `+=`, sehingga diskon bisa terakumulasi dari lebih dari satu syarat sekaligus.
7. Baris `nominal_diskon_3026 = total_belanja_3026 * (total_diskon_persen_3026 / 100)` menghitung nilai diskon dalam rupiah, dengan mengubah persentase menjadi pecahan desimal lalu mengalikannya dengan total belanja.
8. Baris `total_bayar_3026 = total_belanja_3026 - nominal_diskon_3026` menghitung total yang harus dibayar setelah dikurangi nominal diskon.
9. Bagian output menggunakan `print()` dengan format string (f-string) untuk menampilkan rincian pembayaran, termasuk format `{:,.0f}` yang menampilkan angka dengan pemisah ribuan dan tanpa angka desimal.

### 2.2.5 Program "if_elif_else1_2611533026.py"
Program ini dibuat untuk menentukan apakah seseorang boleh mengendarai motor berdasarkan dua faktor, yaitu umur dan kepemilikan SIM C. Pengguna diminta memasukkan umur serta status kepemilikan SIM C, kemudian program akan memeriksa kombinasi dari kedua data tersebut secara berurutan untuk menghasilkan salah satu dari empat kemungkinan pesan. Program ini menerapkan konsep struktur percabangan (*conditional statement*) dalam bahasa pemrograman Python, khususnya penggunaan pernyataan `if-elif-else` yang dikombinasikan dengan operator logika `and`, untuk menentukan alur eksekusi program berdasarkan beberapa kondisi yang saling eksklusif.
</div>

```python
umur_3026 = int(input("input umur anda : "))
sim_3026 = input("Apakah anda udah punya SIM C: ") [0] # guna [0] -> array -> liat diawal aja sehingga selama "y" diawal akan tetap benar

if umur_3026 >= 17 and sim_3026 == 'y' :
    print("anda sudah dewasa dan boleh bawa motor")

elif umur_3026 >= 17 and sim_3026 != 'y' :
    print("anda sudah dewasa tetapi tidak boleh bawa motor")

elif umur_3026 < 17 and sim_3026 == 'y' :
    print("anda belum cukup umur punya SIM")

else :
    print ("anda belum cukup umur bawa motor")
```

**Penjelasan :**

### 2.2.6 Program "nilai_2611533026.py"
Program ini dibuat untuk mengonversi nilai angka menjadi nilai huruf (grade) sesuai rentang tertentu, yaitu A untuk nilai 81 ke atas, B untuk 70–80, C untuk 60–69, D untuk 50–59, dan E untuk nilai di bawah 50. Pengguna diminta memasukkan sebuah nilai angka, kemudian program akan memeriksanya secara berurutan untuk menentukan grade yang sesuai. Program ini menerapkan konsep struktur percabangan (*conditional statement*) dalam bahasa pemrograman Python, khususnya penggunaan pernyataan `if-elif-else` bertingkat, untuk menangani beberapa kemungkinan kondisi yang saling eksklusif berdasarkan rentang nilai.
</div>

```python
nilai_3026 = int(input("Inputkan nilai angka = "))

if nilai_3026 >= 81 :
    print('A')
elif nilai_3026 >= 70 :
   print('B') 
elif nilai_3026>= 60 :
    print('C')
elif nilai_3026>= 50 :
    print('D')
else :
    print('E')
```

**Penjelasan :**
1. Baris `nilai_3026 = int(input("Inputkan nilai angka = "))` meminta pengguna memasukkan nilai angka, lalu mengubahnya menjadi bilangan bulat dengan `int()`, dan menyimpannya ke variabel `nilai_3026`.
2. Baris `if nilai_3026 >= 81 :` memeriksa apakah nilai yang dimasukkan 81 ke atas. Jika `True`, program mencetak `'A'`.
3. Baris `elif nilai_3026 >= 70 :` hanya diperiksa jika kondisi sebelumnya `False`. Baris ini memeriksa apakah nilai berada di rentang 70 ke atas (tetapi di bawah 81, karena kondisi 81 ke atas sudah tidak terpenuhi). Jika `True`, program mencetak `'B'`.
4. Baris `elif nilai_3026 >= 60 :` hanya diperiksa jika kedua kondisi sebelumnya `False`, sehingga mewakili rentang nilai 60 ke atas (di bawah 70). Jika `True`, program mencetak `'C'`.
5. Baris `elif nilai_3026 >= 50 :` hanya diperiksa jika ketiga kondisi sebelumnya `False`, sehingga mewakili rentang nilai 50 ke atas (di bawah 60). Jika `True`, program mencetak `'D'`.
6. Baris `else :` menangani seluruh kemungkinan sisa, yaitu nilai di bawah 50, dengan mencetak `'E'`.
7. Karena menggunakan `if-elif-else`, Python mengevaluasi kondisi secara berurutan dari atas ke bawah dan langsung berhenti begitu satu kondisi terpenuhi, sehingga hanya **satu** huruf saja yang akan tercetak sebagai hasil akhir.

### 2.2.7 Program "match_case1_2611533026.py"
Program ini dibuat untuk mengonversi angka bulan (1–12) menjadi nama bulan dalam Bahasa Indonesia. Pengguna diminta memasukkan angka bulan, kemudian program akan mencocokkan angka tersebut dengan salah satu dari dua belas kemungkinan nilai yang tersedia, dan menampilkan nama bulan yang sesuai. Jika angka yang dimasukkan berada di luar rentang 1–12, program akan menampilkan pesan bahwa angka tersebut tidak valid. Program ini menerapkan konsep struktur percabangan (conditional statement) dalam bahasa pemrograman Python, khususnya penggunaan pernyataan match-case, sebagai alternatif dari if-elif-else untuk menangani banyak kemungkinan nilai yang bersifat diskret secara lebih ringkas dan terstruktur.
</div>

```python
bulan_3026 = int(input("masukkan angka bulan (1-12) :"))

match bulan_3026 :
    case 1 :
        print("Januari")
    case 2 :
        print("Februari")
    case 3 :
            print("Maret")
    case 4 :
            print("April")
    case 5 :
            print("Mei")
    case 6 :
            print("Juni")
    case 7 :
            print("Juli")
    case 8 :
            print("Agustus")
    case 9 :
            print("September")
    case 10 :
            print("Oktober")
    case 11 :
            print("November")
    case 12 :
            print("Desember")
    case _ :
            print("Angka tidak valid")
```

<div style="font-family: 'Times New Roman', Times, serif; font-size: 18px;">

**Penjelasan :**
1. Baris `bulan_3026 = int(input("masukkan angka bulan (1-12) :"))` meminta pengguna memasukkan angka bulan, lalu mengubahnya menjadi bilangan bulat dengan `int()`, dan menyimpannya ke variabel `bulan_3026`.
2. Baris `match bulan_3026 :` membuka struktur `match-case`, yaitu bentuk percabangan alternatif di Python (mirip `switch-case` pada bahasa lain) yang akan mencocokkan nilai variabel `bulan_3026` dengan salah satu `case` di bawahnya.
3. Setiap baris `case 1 :`, `case 2 :`, dan seterusnya hingga `case 12 :` mewakili satu kemungkinan nilai bulan. Python akan mencocokkan nilai `bulan_3026` dengan angka pada setiap `case` secara berurutan dari atas ke bawah.
4. Begitu ditemukan `case` yang cocok dengan nilai `bulan_3026`, baris `print(...)` di bawah `case` tersebut dijalankan (mencetak nama bulan yang sesuai), lalu Python langsung berhenti memeriksa `case` lainnya.
5. Baris `case _ :` di bagian akhir adalah **wildcard** atau penampung default, yang akan dijalankan jika nilai `bulan_3026` tidak cocok dengan `case` mana pun (misalnya angka di luar 1–12), dengan mencetak pesan bahwa angka tidak valid.
6. Secara konsep, `match-case` di sini bekerja mirip dengan `if-elif-else` bertingkat, hanya saja lebih ringkas dan lebih cocok digunakan ketika suatu variabel perlu dicocokkan dengan banyak nilai yang bersifat pasti (diskret), bukan berupa rentang atau kondisi logika kompleks.

<h1 align="center";>BAB III <br> KESIMPULAN </h1>

Berdasarkan seluruh praktikum yang telah dilaksanakan, dapat disimpulkan bahwa struktur percabangan merupakan salah satu elemen fundamental dalam bahasa pemrograman Python yang memungkinkan suatu program untuk mengambil keputusan berdasarkan kondisi tertentu. Struktur `if` tunggal digunakan ketika hanya terdapat satu kondisi yang perlu diperiksa, sedangkan `if-else` digunakan ketika terdapat dua kemungkinan hasil yang saling melengkapi.

Struktur `if` bertingkat (*nested if*) diterapkan ketika suatu kondisi memiliki sub-kondisi yang perlu diperiksa lebih lanjut, sementara beberapa `if` sejajar yang bersifat independen cocok digunakan pada kasus-kasus yang membutuhkan akumulasi hasil dari beberapa syarat yang dapat terpenuhi secara bersamaan, seperti pada perhitungan diskon belanja. Berbeda halnya dengan `if-elif-else`, yang cocok digunakan pada kondisi yang bersifat saling eksklusif, di mana hanya satu kemungkinan saja yang akan terpenuhi dan dieksekusi, seperti pada penentuan grade nilai maupun kelayakan berkendara.

Selain itu, struktur `match-case` hadir sebagai alternatif yang lebih ringkas dan terstruktur dibandingkan `if-elif-else` bertingkat, khususnya ketika suatu variabel perlu dicocokkan dengan banyak nilai diskret, seperti pada konversi angka bulan menjadi nama bulan.

Melalui studi kasus terintegrasi berupa sistem loket taman wahana, dapat dipahami bahwa dalam praktiknya, sebuah program yang realistis seringkali tidak hanya menggunakan satu jenis struktur percabangan saja, melainkan mengombinasikan beberapa jenis struktur sekaligus sesuai dengan kebutuhan logika masing-masing bagian program. Pemahaman yang baik terhadap karakteristik setiap struktur percabangan akan sangat membantu seorang pemrogram dalam menyusun kode program yang efisien, mudah dibaca, dan sesuai dengan kebutuhan permasalahan yang dihadapi.
