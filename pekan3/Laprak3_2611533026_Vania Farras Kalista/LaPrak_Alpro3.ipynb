### 2.2.1 Program "Numerik_2611533026"
Program Python dibuat menggunakan Visual Studio Code. Pada praktikum ini, Python digunakan untuk membuat dan menjalankan program sederhana sebagai pengenalan terhadap sintaks dasar bahasa Python.

```python
#Buat file dengan nama Numerik_2611533026.py
#Buat variabel tipe interger
a_3026 = 100
print("the type of variable having value", a_3026, "is", type(a_3026))

#Buat variabel tipe float
b_3026 = 10.2345
print("the type of variable having value", b_3026, "is", type(b_3026))

#buat variabel tipe kompleks
c_3026 = 100 + 3j
print("the type of variable having value", c_3026, "is", type(c_3026))

#note : gunanya program ini untuk mengetahui tipe data dari variabel yang telah dibuat
```

3. print() digunakan untuk menampilkan data ke layar.
4. \n digunakan untuk membuat baris baru pada output yang ditampilkan.
