## 2.1 Dasar Teori
### 2.1.1 Operator Aritmatika
Operator aritmatika adalah simbol yang digunakan dalam bahasa pemrograman untuk melakukan operasi perhitungan matematika terhadap satu atau lebih nilai (operand). Dalam Python, operator aritmatika digunakan untuk memanipulasi data numerik seperti bilangan bulat (integer) dan bilangan pecahan (float).

1. Operator Penjumlahan (+) :
Digunakan untuk menjumlahkan dua nilai numerik atau menggabungkan (concatenate) dua string.

2. Operator Pengurangan (-) :
Digunakan untuk mengurangkan nilai kedua dari nilai pertama.

3. Operator Perkalian (*) :
Digunakan untuk mengalikan dua nilai. Operator ini juga bisa digunakan untuk mengulang string, misalnya "ab" * 3 menghasilkan "ababab".

4. Operator Pembagian (/) :
Selalu menghasilkan tipe data float, meskipun hasil pembagiannya adalah bilangan bulat. Contoh: 6 / 2 menghasilkan 3.0, bukan 3.

5. Operator Pembagian Bulat (//) :
Menghasilkan hasil bagi yang dibulatkan ke bawah (ke arah bilangan bulat terdekat yang lebih kecil). Contoh: 7 // 2 menghasilkan 3.

6. Operator Modulus (%) :
Menghasilkan sisa dari pembagian dua bilangan. Sering digunakan untuk mengecek bilangan genap/ganjil, misalnya angka % 2 == 0.

7. Operator Perpangkatan ()** :
Digunakan untuk memangkatkan suatu bilangan. Contoh: 2 ** 4 menghasilkan 16 (2 pangkat 4).

### 2.1.2 Operator Perbandingan
Operator perbandingan digunakan untuk membandingkan dua nilai dan menghasilkan tipe data boolean (True atau False).

1. `==` Sama dengan : 5 == 5 menghasilkan True
2. `!=` Tidak sama dengan : 5 != 3 menghasilkan True
3. `>` Lebih besar dari : 5 > 3 menghasilkan True
4. `<` Lebih kecil dari : 5 < 3 menghasilkan False
5. `>=` Lebih besar sama dengan : 5 >= 5 menghasilkan True
6. `<=` Lebih kecil sama dengan : 5 <= 3 menghasilkan False

Operator ini banyak digunakan dalam struktur kontrol seperti if dan while untuk menentukan alur eksekusi program berdasarkan suatu kondisi.


### 2.1.3 Operator Logika
Operator logika digunakan untuk menggabungkan dua atau lebih ekspresi kondisi (boolean) menjadi satu hasil evaluasi.
1. `and` (Dan) : bernilai True jika kedua kondisi benar. Contoh: True and False menghasilkan False
2. `or` (Atau) : bernilai True jika salah satu kondisi benar. Contoh: True or False menghasilkan True
3. `not` (Negasi) : membalikkan nilai boolean. Contoh: not True menghasilkan False

### 2.1.4 Operator Penugasan
Operator penugasan digunakan untuk memberikan nilai ke suatu variabel, termasuk penugasan yang digabung dengan operasi aritmatika (compound assignment).

1. `=` : penugasan biasa. Contoh : x = 5
2. `+=` : x += 3 setara dengan x = x + 3
3. `-=` : x -= 3 setara dengan x = x - 3
4. `*=` : x *= 3 setara dengan x = x * 3
5. `/=` : x /= 3 setara dengan x = x / 3
6. `//=` : x //= 3 setara dengan x = x // 3
7. `%=` : x %= 3 setara dengan x = x % 3
8. `**=` : x **= 3 setara dengan x = x ** 3

Operator ini mempermudah penulisan kode agar lebih ringkas, terutama saat memperbarui nilai suatu variabel secara berulang

### 2.1.5 Operator Keanggotaan
Operator keanggotaan digunakan untuk memeriksa apakah suatu nilai terdapat di dalam suatu struktur data seperti list, tuple, string, atau dictionary.

1. `in` : mengecek apakah nilai ada di dalam objek. Contoh: "a" in "kata" menghasilkan True
2. `not in` : mengecek apakah nilai tidak ada di dalam objek. Contoh: "z" in "kata" menghasilkan False

### 2.1.6 Operator Identitas
Operator identitas digunakan untuk membandingkan apakah dua variabel merujuk pada objek yang sama di memori (bukan sekadar nilainya sama).

1. `is` : bernilai True jika kedua objek sama (identik di memori)
2. `is not` : bernilai True jika kedua objek berbeda

### 2.1.7 Operator Bitwise
Operator bitwise digunakan untuk melakukan operasi pada level bit (representasi biner) dari suatu bilangan bulat.

1. `&` (AND) : bernilai 1 jika kedua bit bernilai 1. Contoh: 5 & 3 menghasilkan 1
2. `|` (OR) : bernilai 1 jika salah satu bit bernilai 1. Contoh: 5 | 3 menghasilkan 7
3. `^` (XOR) : bernilai 1 jika kedua bit berbeda. Contoh: 5 ^ 3 menghasilkan 6
4. `~` (NOT) : membalikkan seluruh bit (komplemen). Contoh: ~5 menghasilkan -6
5. `<<` (Left Shift) : menggeser bit ke kiri. Contoh: 5 << 1 menghasilkan 10
6. `>>` (Right Shift) : menggeser bit ke kanan. Contoh: 5 >> 1 menghasilkan 2

```python
print("=== SISTEM TRANSAKSI TOKO ===")
print()

# Input data pelanggan
nama_3026 = input("Masukkan Nama Pelanggan : ")
status_3026 = input("Masukkan Status Pelanggan (member/nonmember) : ")
total_belanja_3026 = int(input("Masukkan Total Belanja : "))
jumlah_barang_3026 = int(input("Masukkan Jumlah Barang : "))
kode_promo_3026 = input("Masukkan Kode Promo : ")


# ==========================================
# OPERATOR PERBANDINGAN
# ==========================================

# Membandingkan total belanja dengan batas minimum
syarat_belanja_3026 = total_belanja_3026 >= 200000

# Membandingkan jumlah barang dengan batas minimum
syarat_barang_3026 = jumlah_barang_3026 >= 3

# Membandingkan status pelanggan
status_member_3026 = status_3026 == "member"


# ==========================================
# OPERATOR KEANGGOTAAN
# ==========================================

# List kode promo yang tersedia
daftar_promo_3026 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

# Operator IN
promo_tersedia_3026 = kode_promo_3026 in daftar_promo_3026

# Operator NOT IN
promo_tidak_tersedia_3026 = kode_promo_3026 not in daftar_promo_3026


# ==========================================
# OPERATOR IDENTITAS
# ==========================================

# Mengikuti pola dari materi praktikum
objek1_3026 = daftar_promo_3026
objek2_3026 = objek1_3026
objek3_3026 = daftar_promo_3026.copy()

# Operator IS
identitas_sama_3026 = objek1_3026 is objek2_3026

# Operator IS NOT
identitas_berbeda_3026 = objek1_3026 is not objek3_3026


# ==========================================
# OPERATOR LOGIKA
# ==========================================

# AND: member dan belanja minimal Rp200.000
diskon_member_3026 = status_member_3026 and syarat_belanja_3026

# AND: jumlah barang minimal 3 dan kode promo tersedia
mendapatkan_promo_3026 = syarat_barang_3026 and promo_tersedia_3026

# OR: member atau memiliki promo
akses_pelanggan_3026 = status_member_3026 or promo_tersedia_3026

# NOT: bukan member
bukan_member_3026 = not status_member_3026


# ==========================================
# OPERATOR ARITMATIKA
# ==========================================

# Menghitung diskon 10% untuk member
if diskon_member_3026:
    diskon_3026 = total_belanja_3026 * 10 / 100
else:
    diskon_3026 = 0

# Menghitung total pembayaran
total_pembayaran_3026 = total_belanja_3026 - diskon_3026

# Menghitung rata-rata harga barang
rata_rata_3026 = total_belanja_3026 / jumlah_barang_3026

# Operator modulus
sisa_barang_3026 = jumlah_barang_3026 % 2


# ==========================================
# OPERATOR PENUGASAN
# ==========================================

poin_3026 = 0

# Augmented assignment +=
if diskon_member_3026:
    poin_3026 += 10

# Augmented assignment -=
total_pembayaran_3026 -= 0


# ==========================================
# OPERATOR BITWISE
# ==========================================

# 0001 = Member
# 0010 = Belanja >= Rp200.000
# 0100 = Jumlah Barang >= 3
# 1000 = Kode Promo Tersedia

kode_member_3026 = 1
kode_belanja_3026 = 2
kode_barang_3026 = 4
kode_promo_bit_3026 = 8

# Nilai awal kode status
kode_status_3026 = 0

# Operator OR (|) untuk menggabungkan kondisi
if status_member_3026:
    kode_status_3026 = kode_status_3026 | kode_member_3026

if syarat_belanja_3026:
    kode_status_3026 = kode_status_3026 | kode_belanja_3026

if syarat_barang_3026:
    kode_status_3026 = kode_status_3026 | kode_barang_3026

if promo_tersedia_3026:
    kode_status_3026 = kode_status_3026 | kode_promo_bit_3026


# Operator AND (&)
cek_member_3026 = kode_status_3026 & kode_member_3026
cek_promo_3026 = kode_status_3026 & kode_promo_bit_3026

# Operator XOR (^)
kode_referensi_3026 = 11
perbedaan_status_3026 = kode_status_3026 ^ kode_referensi_3026

# Operator shift <<
kode_shift_3026 = kode_status_3026 << 1


# ==========================================
# HAK AKSES
# ==========================================

member_access_3026 = cek_member_3026 != 0
promo_access_3026 = cek_promo_3026 != 0

# Hak free shipping jika member dan promo tersedia
free_shipping_access_3026 = status_member_3026 and promo_tersedia_3026


# ==========================================
# OUTPUT DATA TRANSAKSI
# ==========================================

print()
print()
print("=== DATA TRANSAKSI ===")

print("Nama Pelanggan       :", nama_3026)
print("Status Pelanggan     :", status_3026)
print("Total Belanja        : Rp" + str(total_belanja_3026))
print("Jumlah Barang        :", jumlah_barang_3026)
print("Kode Promo           :", kode_promo_3026)


# ==========================================
# HASIL VALIDASI
# ==========================================

print()
print()
print("=== HASIL VALIDASI ===")

print("Belanja >= Rp200000        :", syarat_belanja_3026)
print("Jumlah Barang >= 3         :", syarat_barang_3026)
print("Status Member              :", status_member_3026)
print("Kode Promo Tersedia        :", promo_tersedia_3026)
print("Mendapatkan Diskon         :", diskon_member_3026)
print("Mendapatkan Promo          :", mendapatkan_promo_3026)


# ==========================================
# HASIL PERHITUNGAN
# ==========================================

print()
print()
print("=== HASIL PERHITUNGAN ===")

print("Diskon                     : Rp" + str(int(diskon_3026)))
print("Total Pembayaran           : Rp" + str(int(total_pembayaran_3026)))
print("Rata-rata Harga Barang     : Rp" + str(int(rata_rata_3026)))


# ==========================================
# HAK AKSES PELANGGAN
# ==========================================

print()
print()
print("=== HAK AKSES PELANGGAN ===")

print("Kode Hak Akses             :", kode_status_3026)
print("Member Access              :", member_access_3026)
print("Promo Access               :", promo_access_3026)
print("Free Shipping Access       :", free_shipping_access_3026)


# ==========================================
# OPERASI BITWISE
# ==========================================

print()
print()
print("=== OPERASI BITWISE ===")

print("=== Kode Status Transaksi ===")
print()
print("0001 | 0010 | 0100 | 1000")
print()
print("Kode Biner   :", format(kode_status_3026, "b"))
print("Kode Desimal :", kode_status_3026)


# ==========================================
# PEMERIKSAAN STATUS
# ==========================================

print()
print("=== Pemeriksaan Status ===")

print()
print("Cek Member")
print()
print(format(kode_status_3026, "b"), "&", format(kode_member_3026, "04b"))
print()
print("Hasil Biner   :", format(cek_member_3026, "04b"))
print("Hasil Desimal :", cek_member_3026)


print()
print()
print("Cek Promo")
print()
print(format(kode_status_3026, "b"), "&", format(kode_promo_bit_3026, "04b"))
print()
print("Hasil Biner   :", format(cek_promo_3026, "04b"))
print("Hasil Desimal :", cek_promo_3026)


# ==========================================
# PERBANDINGAN STATUS
# ==========================================

print()
print("=== Perbandingan Status ===")
print()
print("Kode Transaksi :", format(kode_status_3026, "b"))
print()
print("Kode Referensi :", format(kode_referensi_3026, "04b"))
print()
print(format(kode_status_3026, "b"), "^", format(kode_referensi_3026, "04b"))
print()
print("Hasil Biner   :", format(perbedaan_status_3026, "04b"))
print("Hasil Desimal :", perbedaan_status_3026)


# ==========================================
# SHIFT
# ==========================================

print()
print("=== Shift ===")
print()
print(format(kode_status_3026, "b"), "<< 1")
print()
print("Hasil Biner   :", format(kode_shift_3026, "b"))
print("Hasil Desimal :", kode_shift_3026)


# ==========================================
# SELESAI
# ==========================================

print()
print("=== SELESAI ===")
```

2. Operator aritmatika (`+`, `-`, `*`, `/`, `//`, `%`, `**`) digunakan untuk melakukan perhitungan matematis, sedangkan operator perbandingan (`==`, `!=`, `>`, `<`, `>=`, `<=`) digunakan untuk membandingkan dua nilai dan menghasilkan tipe data boolean.

3. Operator logika (`and`, `or`, `not`) memungkinkan penggabungan beberapa kondisi boolean sekaligus, sehingga sangat berguna dalam menentukan validasi bertingkat, seperti pengecekan status member dan syarat belanja minimum pada studi kasus sistem transaksi toko.

4. Operator penugasan, khususnya *augmented assignment* (`+=`, `-=`, `*=`, dan sejenisnya), mempermudah penulisan kode agar lebih ringkas dibandingkan penulisan operasi secara manual.

5. Operator keanggotaan (`in`, `not in`) berguna untuk memeriksa keberadaan suatu nilai di dalam struktur data seperti list, sedangkan operator identitas (`is`, `is not`) digunakan untuk membandingkan apakah dua variabel merujuk pada objek yang sama di memori — berbeda dengan operator `==` yang hanya membandingkan nilai/isi.

6. Operator bitwise (`&`, `|`, `^`, `~`, `<<`, `>>`) terbukti dapat diterapkan pada kasus nyata, seperti penentuan hak akses pelanggan melalui teknik *bit flag*, di mana setiap kondisi direpresentasikan oleh satu bit tertentu dan digabungkan menggunakan operator OR, kemudian diperiksa kembali menggunakan operator AND.

7. Melalui studi kasus sistem transaksi toko, seluruh jenis operator yang telah dipelajari dapat diintegrasikan secara nyata dalam satu program yang kompleks, mulai dari validasi data pelanggan, perhitungan diskon dan total pembayaran, hingga penentuan hak akses berdasarkan kombinasi beberapa kondisi.

8. Praktikum ini memberikan pemahaman bahwa pemilihan operator yang tepat sangat memengaruhi efisiensi, keterbacaan, dan kebenaran logika suatu program, sehingga penguasaan terhadap seluruh jenis operator menjadi fondasi penting sebelum mempelajari materi pemrograman Python yang lebih lanjut.
