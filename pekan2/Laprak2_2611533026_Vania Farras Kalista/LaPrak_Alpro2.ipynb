## 2.1 Dasar Teori
### 2.1.1 Python
Python adalah bahasa pemrograman tingkat tinggi yang bersifat general-purpose, artinya dapat digunakan untuk berbagai keperluan pengembangan, mulai dari aplikasi web, analisis data, kecerdasan buatan, hingga otomasi tugas. Python dikembangkan dengan filosofi readability, sehingga strukturnya menyerupai bahasa manusia dan mudah dipahami oleh pemula.

### 2.1.2 string
String adalah tipe data yang digunakan untuk menyimpan dan mengelola teks. String merupakan kumpulan karakter yang disusun secara berurutan dan diperlakukan sebagai satu kesatuan data, di mana karakternya bisa berupa huruf, angka, simbol, maupun spasi. Contoh data string antara lain nama pengguna, kalimat, alamat email, dan kata sandi. 

Beberapa karakteristik penting string:
1. Terdiri dari kumpulan karakter dan memiliki panjang (length) tertentu 
2. Setiap karakter dapat diakses menggunakan indeks, mirip seperti array 
3. Secara konsep, string sering direpresentasikan sebagai array karakter, misalnya string "DATA" memiliki indeks D(0), A(1), T(2), A(3) 

Dalam Python, string dituliskan menggunakan tanda kutip tunggal ('...') atau ganda ("..."), contoh: nama = "Budi". Operasi dasar yang umum dilakukan pada string meliputi mengakses karakter berdasarkan indeks, penggabungan (concatenation), menghitung panjang string, membandingkan dua string, pencarian karakter/kata, serta pemotongan bagian string (substring/slicing).


### 2.1.3 Boolean
Boolean adalah tipe data logika yang hanya memiliki dua kemungkinan nilai: benar atau salah. Dalam Python, nilai boolean ditulis sebagai True dan False (huruf awal kapital), dan tipe datanya disebut bool.

Boolean umumnya dihasilkan dari operasi perbandingan (misalnya 5 > 3 menghasilkan True) atau digunakan sebagai kondisi dalam percabangan (if) dan perulangan. Secara konsep umum pada bahasa pemrograman, setiap nilai juga dapat dievaluasi secara implisit sebagai benar atau salah. nilai seperti angka nol, string kosong, atau None/null umumnya dianggap salah sedangkan nilai-nilai lainnya dianggap benar. Boolean sangat penting dalam pengambilan keputusan di dalam program.

### 2.1.4 Numerik
Python menyediakan tiga tipe data numerik utama:
1. Integer (int) — bilangan bulat, baik positif maupun negatif, tanpa angka desimal. Contoh: 10, -5, 1000.
2. Float (float) — bilangan pecahan/desimal (floating point). Contoh: 3.14, -0.5, 2.0.
3. Complex (complex) — bilangan kompleks yang terdiri dari bagian real dan imajiner, ditulis dengan akhiran j. Contoh: 3+5j, 2-4j.

Ketiga tipe ini dapat dicek menggunakan fungsi bawaan type(), misalnya type(10) akan menghasilkan <class 'int'>. Python juga mendukung konversi antar tipe numerik (casting), seperti int(), float(), dan complex(), serta operasi aritmetika standar (+, -, *, /, //, %, **) yang berlaku pada tipe-tipe ini.

### 2.1.5 Fungsi Input
Fungsi input() digunakan untuk mengizinkan pengguna memasukkan data (input) ke dalam program.
Sintaksnya adalah: input(prompt) di mana parameter prompt berupa string yang merepresentasikan pesan yang ditampilkan sebelum pengguna memasukkan data. Parameter ini bersifat opsional.

Perlu dicatat bahwa nilai yang dikembalikan oleh input() selalu bertipe string, sehingga jika ingin diolah sebagai angka, hasilnya perlu dikonversi terlebih dahulu menggunakan int() atau float().

### 2.1.6 Konstanta
Konstanta (atau nilai konstan) adalah sebuah variabel yang nilainya dideklarasikan di awal dan tidak bisa diubah setelahnya.Python sendiri sebenarnya tidak memiliki mekanisme konstanta bawaan secara native seperti bahasa lain (misalnya const di JavaScript). Namun, deklarasi konstanta di Python dapat dilakukan dengan bantuan tipe dari modul typing, yaitu Final, yang perlu diimpor terlebih dahulu.

Final hanya memberikan informasi ke type checker (seperti mypy/Pylance) bahwa suatu variabel bersifat konstan — pada saat program dijalankan (runtime), Python tidak akan menghasilkan error meskipun nilai variabel yang ditandai Final tersebut diubah, karena sifat immutable ini hanya berlaku di level pengecekan tipe, bukan di runtime. 

Sebagai konvensi, berdasarkan panduan gaya penulisan kode PEP 8, nama konstanta di Python dituliskan dalam huruf besar semua (UPPER_CASE), misalnya PI, MAX_VALUE, TOTAL_MONTH

### 2.2.5 Program "Konstanta_2611533026"
Program Python dibuat menggunakan Visual Studio Code. Pada praktikum ini, Python digunakan untuk membuat dan menjalankan program sederhana bertema perhitungan luas lingkaran, guna mengenal penggunaan konstanta (Final), fungsi input() yang dikonversi ke tipe float, serta format string menggunakan operator %.

```python
#Program ini menggunakan konstanta untuk menghitung luas lingkaran

from typing import Final
PI: Final[float] = 3.14  # konstanta phi
print("pi: %f" % (PI))
jari_3026 = float(input('Masukkan nilai jari-jari:'))
luas_3026 = PI * jari_3026* jari_3026
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3026, luas_3026))
```

**Penjelasan :**
1. `#Program ini menggunakan konstanta untuk menghitung luas lingkaran` digunakan sebagai komentar satu baris dalam bahasa Python untuk memberi keterangan mengenai isi program, dan tidak akan dieksekusi oleh interpreter.
2. `from typing import Final` digunakan untuk meng-*import* tipe `Final` dari modul `typing`, yang merupakan bagian dari Python standard library. Tipe `Final` ini digunakan untuk menandai suatu variabel sebagai **konstanta**.
3. `PI: Final[float] = 3.14  # konstanta phi` digunakan untuk membuat konstanta bernama `PI` dengan tipe data `float` dan nilai `3.14`. Penandaan `Final[float]` memberi tahu type checker bahwa nilai `PI` bertipe float dan tidak boleh diubah selama program berjalan. Di bagian akhir baris terdapat komentar `# konstanta phi` yang menjelaskan bahwa variabel ini merepresentasikan nilai konstanta matematika π (phi).
4. `print("pi: %f" % (PI))` digunakan untuk menampilkan nilai konstanta `PI` ke layar. Tanda `%f` merupakan format specifier yang digunakan untuk menampilkan nilai bertipe float, sedangkan operator `%` di luar tanda kutip digunakan untuk menyisipkan nilai `PI` ke posisi `%f` tersebut.
5. `jari_3026 = float(input('Masukkan nilai jari-jari:'))` digunakan untuk menampilkan pesan `"Masukkan nilai jari-jari:"` dan meminta pengguna memasukkan data. Karena hasil dari `input()` selalu bertipe string, fungsi `float()` digunakan untuk mengonversi (*casting*) nilai tersebut menjadi tipe **float**, agar dapat digunakan dalam operasi perhitungan matematika. Hasilnya disimpan ke variabel `jari_3026`.
6. `luas_3026 = PI * jari_3026* jari_3026` digunakan untuk menghitung luas lingkaran menggunakan rumus luas lingkaran (π × r × r), yaitu mengalikan konstanta `PI` dengan variabel `jari_3026` sebanyak dua kali. Hasil perhitungan disimpan ke variabel `luas_3026`.
7. `print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3026, luas_3026))` digunakan untuk menampilkan hasil perhitungan ke layar. Format specifier `%.2f` digunakan untuk menampilkan nilai float dengan pembulatan sebanyak 2 angka di belakang koma, dan operator `%` di luar tanda kutip digunakan untuk menyisipkan nilai `jari_3026` dan `luas_3026` secara berurutan ke dalam teks tersebut.

```python
from typing import Final
# Konstanta batas kelulusan
BATAS_LULUS: Final[float] = 75.0

print("=== SISTEM REGISTRASI PRAKTIKAN ALPRO 2026 ===")

# Input data praktikan
nama_3026 = input("Masukkan Nama Mahasiswa : ")
jenis_kelamin_3026 = input("Masukkan Jenis Kelamin (L/P): ")
umur_3026 = int(input("Masukkan Umur : "))
nilai_3026 = float(input("Masukkan Skor Tes Awal : "))

# Alamat domisili multiline
alamat_3026 = "Griya insani, Jl Durian tarung, Padang"

# Token identifikasi berupa bilangan kompleks
token_3026 = 100 + 3j

# Evaluasi status kelulusan
lulus_3026 = nilai_3026 >= BATAS_LULUS

print("\n=== DATA PRAKTIKAN & HASIL PEMERIKSAAN ===")

print("Nama Mahasiswa :", nama_3026, "| Tipe:", type(nama_3026))

print("Alamat Domisili:")
print(alamat_3026, "| Tipe:", type(alamat_3026))

print("Umur :", umur_3026, "tahun | Tipe:", type(umur_3026))
print("Skor Tes Awal :", nilai_3026, "| Tipe:", type(nilai_3026))
print("ID Token Sinyal:", token_3026, "| Tipe:", type(token_3026))

print("\n=== STATUS KELULUSAN PRAKTIKUM ===")
print("Batas Minimum Nilai:", BATAS_LULUS)
print("Apakah Dinyatakan Lulus?:", lulus_3026, "| Tipe:", type(lulus_3026))
```


**Penjelasan :**
1. `from typing import Final` digunakan untuk meng-*import* tipe `Final` dari modul `typing`, yang digunakan untuk mendeklarasikan **konstanta** dalam Python.
2. `# Konstanta batas kelulusan` merupakan komentar satu baris yang menjelaskan tujuan dari variabel di baris berikutnya, dan tidak dieksekusi oleh interpreter.
3. `BATAS_LULUS: Final[float] = 75.0` digunakan untuk membuat konstanta bernama `BATAS_LULUS` bertipe `float` dengan nilai `75.0`, yang akan dijadikan acuan batas minimum kelulusan. Penulisan nama dengan huruf besar semua (`UPPER_CASE`) merupakan konvensi penamaan konstanta pada Python.
4. `print("=== SISTEM REGISTRASI PRAKTIKAN ALPRO 2026 ===")` digunakan untuk menampilkan teks judul/header program ke layar.
5. `# Input data praktikan` merupakan komentar yang menjelaskan bahwa baris-baris di bawahnya berkaitan dengan proses input data.
6. `nama_3026 = input("Masukkan Nama Mahasiswa : ")` digunakan untuk meminta pengguna memasukkan nama, dan hasilnya disimpan ke variabel `nama_3026` dengan tipe data **string**.
7. `jenis_kelamin_3026 = input("Masukkan Jenis Kelamin (L/P): ")` digunakan untuk meminta pengguna memasukkan jenis kelamin, dan hasilnya disimpan sebagai tipe **string**.
8. `umur_3026 = int(input("Masukkan Umur : "))` digunakan untuk meminta pengguna memasukkan umur. Karena hasil `input()` selalu bertipe string, fungsi `int()` digunakan untuk langsung mengonversi (*casting*) nilai tersebut menjadi tipe **integer** sebelum disimpan ke variabel `umur_3026`.
9. `nilai_3026 = float(input("Masukkan Skor Tes Awal : "))` digunakan untuk meminta pengguna memasukkan skor tes awal, kemudian mengonversinya menjadi tipe **float** menggunakan fungsi `float()`.
10. `# Alamat domisili multiline` merupakan komentar yang menjelaskan tujuan dari variabel di baris berikutnya.
11. `alamat_3026 = "Griya insani, Jl Durian tarung, Padang"` digunakan untuk membuat variabel bertipe **string** yang berisi teks alamat domisili.
12. `# Token identifikasi berupa bilangan kompleks` merupakan komentar yang menjelaskan tujuan dari variabel di baris berikutnya.
13. `token_3026 = 100 + 3j` digunakan untuk membuat variabel bernama `token_3026` dengan tipe data **complex**, terdiri dari bagian real (`100`) dan bagian imajiner (`3j`).
14. `# Evaluasi status kelulusan` merupakan komentar yang menjelaskan tujuan dari variabel di baris berikutnya.
15. `lulus_3026 = nilai_3026 >= BATAS_LULUS` digunakan untuk membandingkan `nilai_3026` dengan konstanta `BATAS_LULUS` menggunakan operator `>=`. Hasil perbandingan ini berupa nilai **boolean** (`True` atau `False`) yang disimpan ke variabel `lulus_3026`.
16. `print("\n=== DATA PRAKTIKAN & HASIL PEMERIKSAAN ===")` digunakan untuk menampilkan teks header ke layar. Karakter `\n` di awal teks digunakan untuk membuat baris kosong baru sebelum teks ditampilkan, agar output lebih rapi.
17. `print("Nama Mahasiswa :", nama_3026, "| Tipe:", type(nama_3026))` digunakan untuk menampilkan nama beserta tipe datanya. Fungsi `type()` digunakan untuk memeriksa dan menampilkan tipe data dari suatu variabel.
18. `print("Alamat Domisili:")` dan `print(alamat_3026, "| Tipe:", type(alamat_3026))` digunakan untuk menampilkan label "Alamat Domisili" pada satu baris, kemudian isi alamat beserta tipe datanya pada baris berikutnya.
19. `print("Umur :", umur_3026, "tahun | Tipe:", type(umur_3026))` digunakan untuk menampilkan umur beserta satuan "tahun" dan tipe datanya.
20. `print("Skor Tes Awal :", nilai_3026, "| Tipe:", type(nilai_3026))` digunakan untuk menampilkan skor tes awal beserta tipe datanya.
21. `print("ID Token Sinyal:", token_3026, "| Tipe:", type(token_3026))` digunakan untuk menampilkan nilai token (bilangan kompleks) beserta tipe datanya.
22. `print("\n=== STATUS KELULUSAN PRAKTIKUM ===")` digunakan untuk menampilkan header baru pada baris kosong berikutnya.
23. `print("Batas Minimum Nilai:", BATAS_LULUS)` digunakan untuk menampilkan nilai konstanta `BATAS_LULUS` ke layar.
24. `print("Apakah Dinyatakan Lulus?:", lulus_3026, "| Tipe:", type(lulus_3026))` digunakan untuk menampilkan status kelulusan (`True`/`False`) beserta tipe datanya, yaitu **boolean**.

# KESIMPULAN

Berdasarkan seluruh praktikum yang telah dilakukan, dapat disimpulkan beberapa hal sebagai berikut:

1. Python bersifat ***dynamically typed***, artinya tipe data suatu variabel tidak perlu dideklarasikan secara eksplisit, melainkan otomatis ditentukan berdasarkan nilai yang diberikan saat variabel dibuat. Hal ini dapat diverifikasi menggunakan fungsi bawaan `type()`.

2. Python menyediakan beberapa tipe data dasar yang masing-masing memiliki fungsi dan karakteristik berbeda:
   - String digunakan untuk menyimpan data berupa teks, seperti nama, alamat, dan jenis kelamin.
   - Boolean (`True`/`False`) digunakan untuk merepresentasikan nilai logika, umumnya dihasilkan dari operasi perbandingan dan digunakan dalam pengambilan keputusan melalui struktur kondisi `if`.
   - Numerik (integer, float, dan complex) digunakan untuk menyimpan dan mengolah data berupa angka, baik bilangan bulat, desimal, maupun bilangan kompleks.

3. Fungsi `input()` memungkinkan program menerima data secara interaktif dari pengguna. Namun, nilai yang dihasilkan `input()` selalu bertipe **string**, sehingga apabila data tersebut akan digunakan dalam operasi matematika, diperlukan proses konversi tipe data (*casting*) menggunakan fungsi seperti `int()` atau `float()`.

4. **Konstanta** dalam Python dapat dideklarasikan dengan bantuan `Final` dari modul `typing`. Meskipun Python tidak melarang perubahan nilai konstanta pada saat runtime, penandaan `Final` tetap berguna sebagai informasi bagi type checker dan sebagai konvensi penulisan kode yang baik (nama konstanta ditulis dengan huruf besar/`UPPER_CASE`), sesuai standar PEP 8.

5. Pada program terakhir, seluruh tipe data dan konsep tersebut (string, boolean, integer, float, complex, konstanta, dan `input()`) berhasil digabungkan dalam satu program sistem registrasi praktikan yang lebih kompleks, yang menunjukkan bahwa pemahaman terhadap tipe data dan variabel merupakan fondasi penting sebelum membangun program yang lebih besar dan terstruktur.
