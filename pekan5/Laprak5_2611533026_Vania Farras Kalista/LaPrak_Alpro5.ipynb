## 2.1 Dasar Teori
### 2.1.1 Perulangan (Looping)
Perulangan adalah struktur kontrol yang membuat program menjalankan sekumpulan instruksi secara berulang selama kondisi tertentu terpenuhi atau selama masih ada elemen yang perlu diproses. Dengan perulangan, kode yang sama tidak perlu ditulis berkali-kali sehingga program menjadi lebih ringkas, mudah dibaca, dan mudah dipelihara. Python menyediakan dua jenis perulangan utama, yaitu for dan while.

### 2.1.2 Perulangan for
Perulangan for di Python digunakan untuk mengiterasi (menelusuri) elemen-elemen dari suatu objek yang iterable, seperti list, tuple, string, dictionary, set, dan range(). Berbeda dengan bahasa seperti C atau Java yang memakai inisialisasi, kondisi, dan increment, for di Python bekerja dengan mengambil elemen satu per satu dari urutan sampai semua elemen habis dibaca.

```python
for variabel in iterable:
    # blok kode yang diulang
```
Keterangan:
- `variabel` menampung nilai elemen pada setiap iterasi.
- `iterable` adalah objek yang dapat ditelusuri.
- Blok kode harus menjorok (indentasi) karena Python menggunakan indentasi untuk menandai isi blok.

### 2.1.3 Fungsi range()
Fungsi `range()` menghasilkan deret bilangan bulat yang sering dipakai sebagai iterable pada `for`.

### 2.1.4 Perulangan Bersarang (Nested for)
Nested for adalah perulangan `for` yang berada di dalam perulangan `for` lainnya. Perulangan di bagian dalam (*inner loop*) akan dijalankan sampai selesai pada setiap satu kali iterasi perulangan di bagian luar (*outer loop*). Jika outer loop berjalan sebanyak *m* kali dan inner loop sebanyak *n* kali, maka blok kode paling dalam dieksekusi sebanyak *m × n* kali.

Perulangan bersarang umum digunakan untuk:
- Mencetak pola atau bentuk (segitiga, persegi, piramida).
- Membuat tabel, misalnya tabel perkalian.
- Memproses data dua dimensi seperti matriks atau list di dalam list.

### 2.2.3 Program "perulangan_for3_2611533026.py"
Program ini dibuat untuk mempraktikkan perulangan `for` yang dipadukan dengan variabel penampung (akumulator) dan percabangan `if-else`. Pengguna memasukkan jumlah perulangan, lalu program menampilkan proses penjumlahan bilangan dari 1 sampai angka tersebut dalam bentuk rangkaian, misalnya `1 + 2 + 3 = 6`, dan menampilkan hasil totalnya di akhir. Dengan program ini, praktikan dapat memahami cara menjumlahkan nilai secara bertahap di dalam perulangan menggunakan operator `+=`, serta cara memakai `if-else` untuk mengatur tampilan tanda `+` dan `=`.
</div>

````python
ulang_3026 = int(input("Masukkan jumlah perulangan: "))

jumlah_3026 = 0
for i_3026 in range(1, ulang_3026 + 1):
    print(i_3026, end=" " )
    jumlah_3026 += i_3026

    if i_3026 < ulang_3026:
        print("+", end=" ")

    else:
        print("=", jumlah_3026)

print()
print("Jumlah =", jumlah_3026)
````

   - Angkanya disimpan dalam variabel `ulang_3026`.

2. `jumlah_3026 = 0`
   - Membuat variabel `jumlah_3026` dengan nilai awal 0.
   - Variabel ini menampung total penjumlahan, jadi harus dimulai dari 0.

3. `for i_3026 in range(1, ulang_3026 + 1):`
   - `range(1, ulang_3026 + 1)` menghasilkan deret angka dari 1 sampai `ulang_3026`.
   - Variabel `i_3026` menampung angka tersebut satu per satu di setiap putaran.

4. `print(i_3026, end=" " )`
   - Mencetak angka saat ini.
   - `end=" "` membuat hasil cetak tetap berada di baris yang sama, dipisahkan spasi.

5. `jumlah_3026 += i_3026`
   - Menambahkan angka saat ini ke `jumlah_3026`.
   - Perintah ini sama dengan `jumlah_3026 = jumlah_3026 + i_3026`.

6. `if i_3026 < ulang_3026:`
   - Memeriksa apakah angka saat ini masih lebih kecil dari angka terakhir.
   - Jika ya, berarti masih ada angka berikutnya, sehingga program mencetak tanda `+` dengan `print("+", end=" ")`.

7. `else:`

### 2.2.4 Program "jumlah_genap_2611533026.py"
Program ini dibuat untuk mempraktikkan perulangan `for` yang dipadukan dengan percabangan `if` bersarang dan operator modulus (`%`). Pengguna memasukkan nilai batas, lalu program memilih hanya bilangan genap dari 1 sampai batas tersebut, menampilkannya dalam bentuk rangkaian penjumlahan, dan menampilkan hasil totalnya di akhir. Dengan program ini, praktikan dapat memahami cara menyaring angka berdasarkan kondisi tertentu di dalam perulangan, serta cara menjumlahkan nilai secara bertahap menggunakan operator `+=`.
</div>

````python
ulang_3026 = int(input("Masukkan nilai batas: "))

jumlah_3026 = 0
for i_3026 in range(1, ulang_3026 + 1):
    if i_3026 % 2 == 0:
        print(i_3026, end=" " )
        jumlah_3026 += i_3026

        if i_3026 < ulang_3026:
            print("+", end=" ")

        else:
            print("=", jumlah_3026)

print()
print("Jumlah =", jumlah_3026)
````

2. `jumlah_3026 = 0`
   - Membuat variabel `jumlah_3026` dengan nilai awal 0.
   - Variabel ini menampung total penjumlahan.

   - `range(1, ulang_3026 + 1)` menghasilkan deret angka dari 1 sampai `ulang_3026`.
   - Variabel `i_3026` menampung angka tersebut satu per satu di setiap putaran.

4. `if i_3026 % 2 == 0:`
   - Operator `%` menghitung sisa bagi. Jika sisa bagi 2 sama dengan 0, berarti angkanya genap.
   - Hanya angka genap yang masuk ke blok di bawahnya. Angka ganjil dilewati.

5. `print(i_3026, end=" " )`
   - Mencetak angka genap tersebut.
   - `end=" "` membuat hasil cetak tetap berada di baris yang sama, dipisahkan spasi.

6. `jumlah_3026 += i_3026`
   - Menambahkan angka genap saat ini ke `jumlah_3026`.
   - Perintah ini sama dengan `jumlah_3026 = jumlah_3026 + i_3026`.

7. `if i_3026 < ulang_3026:`
   - Memeriksa apakah angka saat ini masih lebih kecil dari nilai batas.
   - Jika ya, program mencetak tanda `+` dengan `print("+", end=" ")`.

8. `else:`
   - Dijalankan jika angka saat ini sama dengan nilai batas.
   - `print("=", jumlah_3026)` mencetak tanda sama dengan beserta hasil total, lalu pindah baris.

9. `print()`
   - Baris ini tidak menjorok, jadi berada di luar perulangan dan hanya dijalankan sekali.


Hal ini terjadi karena angka genap terakhir (8) lebih kecil dari batas (9), sehingga kondisi `i_3026 < ulang_3026` masih benar dan program mencetak `+`, bukan `=`. Total tetap benar pada baris `Jumlah =`.

### 2.2.5 Program "nested_for1_2611533026.py"
Program ini dibuat untuk mempraktikkan perulangan bersarang (*nested for*) dalam membentuk pola. Pengguna memasukkan nilai batas, lalu program mencetak sejumlah baris, di mana setiap baris berisi deretan titik yang jumlahnya makin berkurang diikuti nomor barisnya. Dengan program ini, praktikan dapat memahami cara kerja outer loop dan inner loop, hubungan antara variabel perulangan luar dengan batas perulangan dalam, serta pengaturan keluaran menggunakan end=" " agar titik dan angka tercetak dalam satu baris.
</div>

````python
batas_3026 = int(input("Masukkan nilai batas: "))
for line_3026 in range(1, batas_3026 + 1):
    for column_3026 in range(1, (-1 * line_3026 + batas_3026 )+ 1):
        print(".", end=" ")
    print(line_3026)
````

<div style="font-family: 'Times New Roman', Times, serif; font-size: 18px;">

**Penjelasan :**

1. `batas_3026 = int(input("Masukkan nilai batas: "))`
   - `input()` meminta pengguna mengetik angka, lalu `int()` mengubahnya menjadi bilangan bulat.
   - Angkanya disimpan dalam variabel `batas_3026`, yang menentukan jumlah baris yang akan dicetak.

2. `for line_3026 in range(1, batas_3026 + 1):`
   - Ini adalah perulangan luar (*outer loop*) yang mengatur baris.
   - `range(1, batas_3026 + 1)` menghasilkan angka dari 1 sampai `batas_3026`, sehingga setiap angka mewakili satu baris.

3. `for column_3026 in range(1, (-1 * line_3026 + batas_3026) + 1):`
   - Ini adalah perulangan dalam (*inner loop*) yang mengatur titik di setiap baris.
   - Ekspresi `-1 * line_3026 + batas_3026` sama dengan `batas_3026 - line_3026`, jadi `range()` menghasilkan angka dari 1 sampai `batas_3026 - line_3026`.
   - Artinya jumlah titik di setiap baris adalah `batas_3026 - line_3026`. Semakin besar nomor baris, semakin sedikit titiknya.
   - Inner loop dijalankan sampai selesai setiap kali outer loop berputar satu kali.

4. `print(".", end=" ")`
   - Baris ini menjorok di dalam inner loop, jadi dijalankan di setiap putaran inner loop.
   - Mencetak satu tanda titik, dan `end=" "` membuat hasil cetak tetap berada di baris yang sama, dipisahkan spasi.

5. `print(line_3026)`
   - Baris ini hanya menjorok di dalam outer loop, jadi dijalankan sekali setelah inner loop selesai.
   - Mencetak nomor baris di belakang titik-titik, lalu otomatis pindah ke baris baru.

### 2.2.6 Program "nested_for2_2611533026.py"

Pengguna memasukkan nilai batas, lalu program mencetak bintang sebanyak batas tersebut pada setiap baris, dengan jumlah baris yang juga sama dengan batas. ### 2.2.8 Program "nested_for4_2611533026.py"
Program ini dibuat untuk mempraktikkan perulangan bersarang (*nested for*) yang dipadukan dengan percabangan `if-elif-else` untuk membentuk pola karakter yang kompleks. Pengguna memasukkan tinggi pola berupa bilangan genap, lalu program memvalidasinya dan mencetak pola berbingkai yang berisi tanda `<>` yang bergerak menjauhi lalu mendekati dinding, membentuk bangun menyerupai jam pasir. Dengan program ini, praktikan dapat memahami cara outer loop mengatur baris dan inner loop mengatur kolom, penggunaan variabel pengatur posisi yang nilainya berubah di setiap baris, serta cara menentukan karakter yang dicetak berdasarkan posisi baris dan kolom.
</div>

````python
tinggi_3026 = int(input("masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3026 % 2 != 0:
    print("Masukkan bilangan genap!!")
else:
    a_3026 = tinggi_3026 
    c_3026 = a_3026
    lebar_3026 = (2*tinggi_3026) - 2

    for i_3026 in range(1, tinggi_3026 + 1):
        b_3026 = c_3026 + 1

        for j_3026 in range(1, lebar_3026 + 1):

            #Baris atas dan bawah
            if i_3026 == 1 or i_3026 == tinggi_3026:
                if j_3026 == 1 or j_3026 == lebar_3026:
                    print("#", end="")
                else:
                    print("=", end="")

            #Baris isi
            else :
                if j_3026 == 1 or j_3026 == lebar_3026:
                    print("|", end="")
                else:
                    if j_3026 == c_3026 :
                        print("<", end="")
                    elif j_3026 == b_3026 :
                        print(">", end="")
                    elif j_3026 == (lebar_3026 - c_3026) :
                        print("<", end="")
                    elif j_3026 == (lebar_3026 - c_3026 + 1) :
                        print(">", end="")
                    elif j_3026 > b_3026 and j_3026 < (lebar_3026 - c_3026) :
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        #logika asli java
        a_3026 -= 2

        if a_3026 <= 0:
            c_3026 = (-a_3026) + 2
        else:
            c_3026 = a_3026
````

   - `lebar_3026 = (2*tinggi_3026) - 2` menentukan lebar pola, yaitu jumlah karakter per baris. Untuk tinggi 10, lebarnya 18.

   - Ini adalah perulangan luar (*outer loop*) yang mengatur baris, dari baris 1 sampai baris terakhir.

   - Jika sedang di baris pertama atau baris terakhir, kolom paling kiri dan paling kanan dicetak `#`.
   - Kolom di antaranya dicetak `=`.

     - `j_3026 == (lebar_3026 - c_3026 + 1)` mencetak `>`.

     - Selain itu mencetak spasi.
   - Semua `print()` memakai `end=""` agar karakter menyambung tanpa spasi dalam satu baris.

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa perulangan for pada Python digunakan untuk menjalankan blok kode secara berulang dengan menelusuri elemen dari suatu iterable, yang dalam praktikum ini berupa range(). Fungsi range() dapat diatur dengan satu, dua, maupun tiga nilai, dan karena nilai batasnya tidak ikut dihitung, penulisan seperti range(1, n + 1) diperlukan agar angka terakhir ikut tercetak. Perulangan for juga dapat dipadukan dengan variabel penampung, operator +=, percabangan if-else, dan operator modulus untuk menjumlahkan serta menyaring data. Pada perulangan bersarang (*nested for*), inner loop dijalankan sampai selesai setiap kali outer loop berputar satu kali, di mana outer loop umumnya mengatur baris dan inner loop mengatur kolom. Dengan membuat batas inner loop bergantung pada variabel outer loop, berbagai pola seperti persegi, segitiga, tabel penjumlahan, dan jam pasir dapat dibentuk. Selain itu, penggunaan end="" atau end=" " pada print() menjaga keluaran tetap satu baris, sedangkan print() kosong berfungsi pindah ke baris berikutnya. Oleh karena itu, ketelitian dalam menentukan batas range() dan menjaga indentasi sangat penting agar program menghasilkan keluaran yang sesuai.
