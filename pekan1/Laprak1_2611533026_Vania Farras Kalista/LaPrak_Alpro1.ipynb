## 2.1 Dasar Teori
### 1. Python
Python adalah bahasa pemrograman tingkat tinggi yang bersifat general-purpose, artinya dapat digunakan untuk berbagai keperluan pengembangan, mulai dari aplikasi web, analisis data, kecerdasan buatan, hingga otomasi tugas. Python dikembangkan dengan filosofi readability, sehingga strukturnya menyerupai bahasa manusia dan mudah dipahami oleh pemula.

### 2. Visual Studio Code
Visual Studio Code (VS Code) merupakan code editor yang digunakan untuk menulis, mengedit, dan menjalankan kode program. VS Code mendukung berbagai bahasa pemrograman, termasuk Python, serta menyediakan fitur seperti terminal, debugging, dan integrasi dengan Git.

### 3. Git
Git adalah sistem kontrol versi terdistribusi yang digunakan untuk melacak dan mengelola setiap perubahan yang terjadi pada berkas kode program dari waktu ke waktu.

**Elemen-elemen penting dalam Git:**
- **Working Directory** — folder aktif tempat berkas proyek disimpan dan diedit.
- **Staging Area** — area sementara untuk menampung perubahan sebelum disimpan secara permanen ke riwayat proyek.
- **Commit** — proses penyimpanan snapshot perubahan kode beserta pesan penjelas (commit message).
- **Branch** — cabang pengembangan yang memungkinkan pengerjaan fitur atau perbaikan dilakukan secara terpisah dari kode utama.
- **Merge** — proses penggabungan perubahan dari satu branch ke branch lainnya.

### 4. GitHub
GitHub adalah platform hosting berbasis cloud untuk repository Git yang memungkinkan penyimpanan kode secara daring serta mendukung kolaborasi antar-pengembang dari berbagai lokasi.

**Elemen-elemen penting dalam GitHub:**
- **Remote Repository** — repositori yang tersimpan di server GitHub dan dapat diakses melalui internet.
- **Push** — proses mengirimkan perubahan dari repositori lokal ke repositori remote di GitHub.
- **Pull** — proses mengambil perubahan terbaru dari repositori remote ke repositori lokal.
- **Clone** — proses menyalin repositori dari GitHub ke komputer lokal.
- **Collaboration Tools** — fitur pendukung kolaborasi seperti Issues, Pull Request, dan Fork.

### 5. Repository
Repository (repo) adalah tempat penyimpanan seluruh berkas proyek beserta riwayat perubahannya yang dikelola menggunakan Git. Repository dapat disimpan secara lokal pada komputer maupun secara daring pada GitHub.

## 2.2 Kode Program

### 2.2.1 Program "B_2611533026"
Program Python dibuat menggunakan Visual Studio Code. Pada praktikum ini, Python digunakan untuk membuat dan menjalankan program sederhana sebagai pengenalan terhadap sintaks dasar bahasa Python.

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 pratikum alpro
"""
print("Hello World")
print()
print("Program ini menghasilkan")
print("Empat baris output")
print("/ \\ // \\\ /// \\\\\ ")
print("program ini" + " dibuat oleh \n"
+ "vania farras kalista dan 2611533026.")
```

**Penjelasan :**

1. '#' digunakan sebagai simbol untuk membuat komentar satu baris dalam bahasa Python.
2. """ digunakan untuk menyimpan string atau komentar yang lebih dari satu baris kode.
3. print() digunakan untuk menampilkan data ke layar.
4. \n digunakan untuk membuat baris baru pada output yang ditampilkan.
5. Program dapat dijalankan melalui terminal atau fitur Run pada VS Code.

### 2.2.2 Membuat Repository GitHub

Setelah program berhasil dibuat dan diuji pada VS Code, langkah selanjutnya adalah menghubungkan proyek tersebut ke GitHub. Hal ini dilakukan agar kode program dapat tersimpan secara daring, sehingga lebih aman dan mudah diakses maupun dikelola kapan saja.

Tahapan yang dilakukan yaitu:

1. Membuat repository baru pada akun GitHub.
2. Membuka folder proyek yang telah dibuat menggunakan VS Code.
3. Membuka terminal yang tersedia pada VS Code.
4. Menginisialisasi Git pada folder proyek menggunakan perintah git init.
5. Menambahkan seluruh file proyek ke dalam staging area menggunakan perintah git add.
6. Membuat commit sebagai penanda perubahan yang telah disimpan, menggunakan perintah git commit.
7. Menghubungkan repository lokal dengan repository GitHub menggunakan perintah git remote add origin.
8. Melakukan push untuk mengunggah kode dari repository lokal ke GitHub menggunakan perintah git push.
